```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_PATH = r"C:\Users\Akshay\OneDrive\Desktop\DMV & ML Practical\DMV Practical\DMV 1\sales_data_sample.csv"
df = pd.read_csv(DATA_PATH, encoding="latin-1")
df
```

```python
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
print("\nDataset Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
df.describe()
```

```python
newdf = df.fillna(0)

print("Missing Values After Cleaning:")
print(newdf.isnull().sum())
```

```python
newdf["LOCATION"] = newdf["CITY"] + ", " + newdf["COUNTRY"]
newdf[["CITY", "COUNTRY", "LOCATION"]].head()
```

```python
total_sales = newdf["SALES"].sum()

print("Total Sales:", total_sales)
```

```python
average_sales = newdf["SALES"].mean()

print("Average Sales:", average_sales)
```

```python
product_sales = newdf.groupby("PRODUCTLINE")["SALES"].sum()

print("Sales by Product Line:")
print(product_sales)
```

```python
product_sales.plot(kind="bar", figsize=(10, 5))

plt.title("Total Sales by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.boxplot(newdf["SALES"])
plt.title("Box Plot of Sales")
plt.ylabel("Sales")
plt.show()
```

```python
plt.figure(figsize=(8, 8))
product_sales.plot(kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("Sales Distribution by Product Line")
plt.ylabel("")
plt.show()
```

```python
newdf.to_csv("cleaned_sales_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```
