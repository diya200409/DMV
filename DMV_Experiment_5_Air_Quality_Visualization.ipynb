```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
DATA_PATH = r"C:\Users\Akshay\OneDrive\Desktop\DMV & ML Practical\DMV Practical\DMV 5\City_Air_Quality.csv"

df = pd.read_csv(DATA_PATH)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
df["Date"] = pd.to_datetime(df["Date"])

print(df.dtypes)
print(df.head())
```

```python
df.describe()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(df["Date"], df["AQI"], marker="o")
plt.title("AQI Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Air Quality Index")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(df["Date"], df["PM2.5"], label="PM2.5")
plt.plot(df["Date"], df["PM10"], label="PM10")
plt.plot(df["Date"], df["CO"], label="CO")

plt.title("Pollutant Levels Over Time")
plt.xlabel("Date")
plt.ylabel("Pollutant Level")
plt.legend()
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
pollutants = df[["PM2.5", "PM10", "CO"]].mean()

plt.figure(figsize=(8, 5))
plt.bar(pollutants.index, pollutants.values)
plt.title("Average Pollutant Levels")
plt.xlabel("Pollutants")
plt.ylabel("Average Level")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["AQI"], bins=10, edgecolor="black")
plt.title("Distribution of AQI Values")
plt.xlabel("AQI")
plt.ylabel("Frequency")
plt.grid(axis="y")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df["AQI"])
plt.title("Boxplot of AQI Values")
plt.xlabel("AQI")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["PM2.5"], df["AQI"])
plt.title("Relationship Between PM2.5 and AQI")
plt.xlabel("PM2.5")
plt.ylabel("AQI")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df[["PM2.5", "PM10", "CO", "AQI"]].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Air Quality Data")
plt.show()
```
